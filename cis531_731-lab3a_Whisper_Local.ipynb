### 1. Load the Model
We will load the `base` model. If you have a GPU and sufficient VRAM, you may experiment with `small` or `medium`.

```python
# This will download the weights (~140MB for 'base') on the first run.
model = whisper.load_model("base")
print("Whisper 'base' model loaded successfully into memory.")
```

### 2. Transcribe Audio Chunk
For this test, we will generate a dummy audio waveform (or you can point `audio_path` to a local `.wav` or `.mp3` file from the stream buffer).

```python
import numpy as np

# SIMULATION: Generate 1 second of white noise audio to verify the pipeline doesn't crash
# In production, replace 'audio_data' with: result = model.transcribe("your_audio_file.wav")
sample_rate = 16000
audio_data = np.random.randn(sample_rate).astype(np.float32)

print("Transcribing stream chunk...")
result = model.transcribe(audio_data)

print("\n--- TRANSCRIPTION OUTPUT ---")
print(result["text"])
print("----------------------------")
```
